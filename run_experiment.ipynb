```python
import subprocess as sp
import pandas as pd
import numpy as np
import os
from io import StringIO
from tqdm import tqdm
from itertools import product
```

```python
dirname = os.environ['GSET'] + '/edwards_anderson/'
filelist = [os.path.join(dirname, f'edwards_anderson_bimodal_l_8_{i}.gset') for i in range(10)]
option_str = '-p {} -r {} --seed {} --pts {} --beta0 {} --beta1 {}'
options = list(product(filelist, [20,40,80], [1], [np.exp(i) for i in range(10)], [0.1], [1.6]))
print(len(options))
```

```python
dflist = []
for opt_comb in tqdm(options):
    arglist = option_str.format(*opt_comb).strip().split()
    output = sp.run(['./pt_icm'] + arglist, stdout=sp.PIPE)
    dflist.append(pd.read_csv(StringIO(output.stdout.decode())))
    
```
