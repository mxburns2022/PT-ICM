```python
import networkx as nx
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
```

```python
true_u = pd.read_csv('./lattice_2d_20_cae_u.csv')
true_c = pd.read_csv('./lattice_2d_20_cae_c.csv')
df = pd.read_csv('results.csv')
relative_err = lambda true, meas: (true-meas) / true
relative_err = np.vectorize(relative_err)
# i_meas,i_truth = np.where(np.isclose(np.array(true_u['T'])[:,None], df['T'][:],atol=0.005))
true_u_vals = np.array(true_u['u'])
true_c_vals = np.array(true_c['c'])
pt_u_vals = np.array(df['u'])
pt_c_vals = np.array(df['C'])
print(pt_c_vals.shape)
print(np.array(df['T']).shape)
comp_df = pd.DataFrame()
comp_df['T'] = true_u['T']
comp_df['$\\varepsilon(T_i)_u$'] = relative_err(true_u_vals, pt_u_vals)
comp_df['$\\varepsilon(T_i)_c$'] = relative_err(true_c_vals, pt_c_vals)
comp_df
```

```python
fig, ax = plt.subplots()
sns.lineplot(ax=ax,data=comp_df, x='T', y='$\\varepsilon(T_i)_u$')
plt.show()
```

```python
fig, ax = plt.subplots()
sns.lineplot(ax=ax,data=comp_df, x='T', y='$\\varepsilon(T_i)_c$')
plt.show()
```

```python
fig, ax = plt.subplots()
sns.lineplot(ax=ax,data=true_c, x='T', y='c')
sns.scatterplot(ax=ax, data=df, x='T', y='C')
plt.show()
```

```python
lattice = nx.grid_2d_graph(m=20, n=20, periodic=True)
lattice:nx.Graph
lattice = nx.convert_node_labels_to_integers(lattice)
```

```python
with open('lattice_2d_20.gset', mode='w') as l50:
    l50.write('{} {}\n'.format(lattice.number_of_nodes(), lattice.number_of_edges()))
    for u,v in lattice.edges():
        l50.write('{} {} {}\n'.format(u, v, -1))
        
```
