```python
import networkx as nx
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
lattice = nx.grid_2d_graph(m=20, n=20, periodic=True)
lattice:nx.Graph
lattice = nx.convert_node_labels_to_integers(lattice)
```

```python
with open('lattice_2d_20.gset', mode='w') as l50:
    l50.write('{} {}\n'.format(lattice.number_of_nodes(), lattice.number_of_edges()))
    for u,v in lattice.edges():
        l50.write('{} {} {}\n'.format(u, v, -1))
        
```
